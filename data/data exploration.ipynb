```python
import pandas as pd
```

```python
courses = pd.read_csv('Online_Courses.csv')
courses
```

```python
courses['Rating'] = (
    courses['Rating']
    .astype(str)                                # convert floats → strings
    .str.replace(r'[^0-9\.]', '', regex=True)   # remove non‑numeric chars
    .replace('', pd.NA)                         # empty → NA
    .astype(float)                               # final numeric
)

courses['Number of viewers'] = (
    courses['Number of viewers']
    .astype(str)
    .str.replace(r'[^0-9]', '', regex=True)
    .replace(['', 'nan', 'None'], pd.NA)
    .astype('Int64')
)


courses
```
